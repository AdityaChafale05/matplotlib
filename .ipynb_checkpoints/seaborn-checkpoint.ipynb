```python
import matplotlib.pyplot as plt 
import pandas as pd
import numpy as np 
import seaborn as sns 
```

```python
sns.set_style("whitegrid")
tips = sns.load_dataset("tips")

print(tips.head())
#Scatter plot 
sns.scatterplot(data=tips , x='total_bill',y='tip', hue = 'time')
plt.show()

# Bar plot with confidence intervals
sns.barplot(data=tips, x='day', y='total_bill', hue='sex')
plt.show()

# Histogram with KDE (Kernel Density Estimate)
sns.histplot(data=tips, x='total_bill', bins=20, kde=True)
plt.show()

# KDE plot alone
sns.kdeplot(data=tips, x='total_bill', hue='time', fill=True)
plt.show()

#box plot
sns.boxplot(data=tips ,x='day', y='total_bill', hue='smoker')
plt.show()

sns.jointplot(data=tips, x='total_bill', y='tip', kind='scatter', hue='time')
plt.show()

```

```python
tips['sex'] = tips['sex'].map({ 'Female': 1 , 'Male':0 })
tips['smoker'] = tips['smoker'].map({ 'No': 0 , 'Yes': 1})
tips['day'] = tips['day'].map({ 'Sun' :1 })
tips['time'] =tips['time'].map({'Dinner': 1  })


corr = tips.corr(numeric_only=True)
print(corr)

sns.heatmap(corr , cmap='plasma'  ,annot=True ,center=0)
plt.show()
```

```python


sns.set_style("whitegrid")
tips = sns.load_dataset("tips")

# 1. Create a figure and a grid of 2 rows and 3 columns
fig, axes = plt.subplots(2, 3, figsize=(18, 10))

# 2. Map each plot to a specific axis (ax)
# Row 0
sns.scatterplot(data=tips, x='total_bill', y='tip', hue='time', ax=axes[0, 0])
sns.barplot(data=tips, x='day', y='total_bill', hue='sex', ax=axes[0, 1])
sns.histplot(data=tips, x='total_bill', bins=20, kde=True, ax=axes[0, 2])

# Row 1
sns.kdeplot(data=tips, x='total_bill', hue='time', fill=True, ax=axes[1, 0])
sns.boxplot(data=tips, x='day', y='total_bill', hue='smoker', ax=axes[1, 1])

# Delete the empty 6th subplot (axes[1, 2]) so it doesn't show a blank box
fig.delaxes(axes[1, 2])

# 3. Final adjustments
plt.tight_layout()
plt.show()

# Note: sns.jointplot creates its own figure and cannot easily be 
# placed inside a standard subplot grid. It should be called separately.
sns.jointplot(data=tips, x='total_bill', y='tip', kind='scatter', hue='time')
plt.show()
```
