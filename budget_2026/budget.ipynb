```python
import matplotlib.pyplot as plt 

#plt.style.use('fast')
years = ['BE 2023-24','BE 2024-25','BE 2025-26', 'BE 2026-27']
expenditure = [4503097,4820512,5065345, 5347315]

values = [162600,172000 ,186454, 219306]


#labeling of data 
plt.ylabel("Expenditure(₹ crore)")
plt.xlabel("Years")
plt.title("Total Government Spending:")


bars1 = plt.bar(years,expenditure,label='Total expenditure',color="skyblue" ,width =0.7)
bars2 = plt.bar(years , values ,label='Defence', color ="navy", width = 0.3)



# Add value labels above bars1
for i, bar in enumerate(bars1):
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, height, 
             f'{height:,}', ha='center', va='bottom')

# Add value labels above bars2
for i, bar in enumerate(bars2):
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, height, 
             f'{height:,}', ha='center', va='bottom')


plt.ylim(0, max(expenditure) * 1.15)
plt.legend(loc='best')
plt.tight_layout()

```

```python
import matplotlib.pyplot as plt 
import numpy as np

# Data
years = ['2023-24', '2024-25', '2025-26', '2026-27']
total_expenditure = [4503097, 4820512, 5065345, 5347315]

plt.figure(figsize=(8,6))

plt.subplot(2,1,1)
plt.plot(years,total_expenditure, label="Total Expenditure",color="gold", linestyle='--',marker='o',)

plt.stackplot(years,total_expenditure, color="gold",alpha=0.3)

plt.title("Overall Government Budget Scale", fontweight='bold')
plt.ylabel("Total (₹ crore)")

for i, val in enumerate(total_expenditure):
    plt.text(i, val, f'{val:,}', va='bottom', ha='center', fontweight='bold', color='orange')

plt.legend(loc="best")
plt.ylim(0, max(expenditure) * 1.15)
plt.grid(True,linestyle='--',alpha=0.3)


# zoomed version 

import matplotlib.patches as mpatches
defence = [162600, 172000, 186454, 219306]
skill = [12892, 6017, 10775, 28715]

colors =['lawngreen','#3498db']

plt.subplot(2,1,2)
patch1 = mpatches.Patch(color='lawngreen', label='Defence', alpha=0.7)
patch2 = mpatches.Patch(color='#3498db', label='Skill', alpha=0.7)
plt.plot(defence, linestyle='--', linewidth=2,color='red', aplha=0.7)
plt.stackplot(years, defence, skill, colors=colors, alpha=0.5)

plt.title("Zoomed view : Defence and Skill & Jobs")
plt.ylabel("Detailed (₹ crore)")

plt.legend(handles=[patch1, patch2], loc="upper left")
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()

for i in range(len(years)):
    plt.text(i, skill[i] + (defence[i]/2), f'{defence[i]:,}', ha='center', color='grey', fontweight='bold', fontsize=9)
    
    plt.text(i, skill[i]/2, f'{skill[i]:,}', ha='center', color='grey', fontweight='bold')
```
